```python
import os
import pandas as pd
```

```python
def rename_files(directory, action):
    # Get list of files in the directory
    files = os.listdir(directory)
    
    # Initialize counter for renaming
    counter = 76
    
    # Iterate over each file
    for file in files:
        # Construct new file name
        new_name = f"{action}{counter}.csv"  # You can adjust the naming pattern
        
        # Rename the file
        os.rename(os.path.join(directory, file), os.path.join(directory, new_name))
        
        print(f"Renamed '{file}' to '{new_name}'")
        
        # Increment counter
        counter += 1

# Example usage
directory_path = "./actual/SHIELD/xy/"
action = "SHIELD"
rename_files(directory_path, action)

```
